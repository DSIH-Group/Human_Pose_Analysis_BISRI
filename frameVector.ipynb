Will construct simplified graph descriptor for each frame with tables and clinicians as nodes and the distances between them as edge weights. Summaries will then be extract from graph to be added to frame vector. Will then combine scenegraph with gpd to make full frame vector






```python
import numpy as np
import pickle
from scene_graph_utils import*

#importing annotations

try:
    with open('gpdAnnots.pkl', 'rb') as f:
        annots = pickle.load(f)
except:
    print("Error")
```

## Computing frame scene graph

```python
for day in annots:
    for operation in annots[day]:
        for frame in annots[day][operation]:
            framePoses = []
            for pose in frame['poses']:
                framePoses.append(pose['keypoints'])

            framePoses = np.array(framePoses)
            frame['graphRep'] = computeSceneGraphSummary(framePoses)
```

```python
graph = annots['day1']['operation_1'][0]['graphRep']
gpd = annots['day1']['operation_1'][0]['meanGpd']
print(graph)
print(gpd)
```

```python
print( annots['day1']['operation_1'][0].keys())
```

## combining into frame vector

```python
for day in annots:
    for operation in annots[day]:
        for frame in annots[day][operation]:
            frame['frameVector'] = np.concatenate([frame['meanGpd'], frame['graphRep']])
```

```python
#inspecting frame vector
fullFrame = annots['day1']['operation_1'][0]['frameVector']
print(f"Frame vector: \n {fullFrame}")
print(f"Shape: {np.shape(fullFrame)}")

```
